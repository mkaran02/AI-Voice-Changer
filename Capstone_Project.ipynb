```python
# ============================================================================
# CELL 6: Download Voice Model
# ============================================================================
#@title **🎤 Step 6: Download Voice Model**
#@markdown Paste your model URL
model_url = "https://huggingface.co/sanjay7178/indian-artists-rvc/resolve/main/Arijit%20Singh%20(RVC)%20(Crepe%20v2)%20450%20Epoch.zip" #@param {type:"string"}
#@markdown Model name (no spaces)
model_name = "arijit_singh" #@param {type:"string"}

import os
import zipfile
import shutil
import urllib.request
import gdown

BASE_DIR = os.getcwd()
rvc_models_dir = os.path.join(BASE_DIR, 'rvc_models')

def extract_zip(extraction_folder, zip_name):
    os.makedirs(extraction_folder, exist_ok=True)
    with zipfile.ZipFile(zip_name, 'r') as zip_ref:
        zip_ref.extractall(extraction_folder)
    os.remove(zip_name)

    index_filepath, model_filepath = None, None
    for root, dirs, files in os.walk(extraction_folder):
        for name in files:
            if name.endswith('.index') and os.stat(os.path.join(root, name)).st_size > 1024 * 100:
                index_filepath = os.path.join(root, name)
            if name.endswith('.pth') and os.stat(os.path.join(root, name)).st_size > 1024 * 1024 * 40:
                model_filepath = os.path.join(root, name)

    if not model_filepath:
        raise Exception(f'No .pth model file found')

    os.rename(model_filepath, os.path.join(extraction_folder, os.path.basename(model_filepath)))
    if index_filepath:
        os.rename(index_filepath, os.path.join(extraction_folder, os.path.basename(index_filepath)))

    for filepath in os.listdir(extraction_folder):
        if os.path.isdir(os.path.join(extraction_folder, filepath)):
            shutil.rmtree(os.path.join(extraction_folder, filepath))

def download_model(url, dir_name):
    try:
        extraction_folder = os.path.join(rvc_models_dir, dir_name)
        if not os.path.exists(extraction_folder):
            print(f'📥 Downloading: {dir_name}...')
            zip_name = url.split('/')[-1]
            urllib.request.urlretrieve(url, zip_name)
            print('📦 Extracting...')
            extract_zip(extraction_folder, zip_name)
            print(f'✅ {dir_name} downloaded!')
        else:
            print(f'⚠️ {dir_name} already exists!')
    except Exception as e:
        print(f"❌ Error: {str(e)}")

def download_online_model_drive(file_id, dir_name):
    try:
        extraction_folder = os.path.join(rvc_models_dir, dir_name)
        if not os.path.exists(extraction_folder):
            print(f'📥 Downloading: {dir_name}...')
            zip_name = f'{dir_name}.zip'
            gdown.download(f'https://drive.google.com/uc?id={file_id}', zip_name, quiet=False)
            print('📦 Extracting...')
            extract_zip(extraction_folder, zip_name)
            print(f'✅ {dir_name} downloaded!')
        else:
            print(f'⚠️ {dir_name} already exists!')
    except Exception as e:
        print(f"❌ Error: {str(e)}")

# Download
if 'drive.google.com' in model_url:
    file_id_drive = model_url.split('/')[-2]
    download_online_model_drive(file_id_drive, model_name)
elif 'huggingface.co' in model_url:
    download_model(model_url, model_name)
else:
    print('❌ Invalid URL')
```

```python
# ============================================================================
# CELL 7: Launch Gradio UI - Option B: Futuristic Dashboard (Neon, 3D, Dark UI)
# ============================================================================
#@title **Step 7: Launch Futuristic Neon Dashboard UI (Option B)**
MURF_API_KEY = "ap2_4f08eadd-fe31-48a1-863e-b082de275cdb" #@param {type:"string"}
rvc_models_dir = '/content/FIX/rvc_models'  # <- update this path if needed

import os, json, requests, subprocess
from pathlib import Path
from datetime import datetime
import gradio as gr

# ------------------------- Utilities ---------------------------------------
def get_available_models():
    try:
        if not os.path.exists(rvc_models_dir):
            return ["No models"]
        return [d for d in os.listdir(rvc_models_dir) if os.path.isdir(os.path.join(rvc_models_dir, d))]
    except Exception:
        return ["No models"]

# ------------------------- Voice conversion --------------------------------
def voice_conversion(audio, model, pitch, idx_rate, flt_rad, rms, prot,
                     mvol, bvol, ivol, pitch_all, rev_sz, rev_wet,
                     rev_dry, rev_damp, fmt, algo):
    """
    Calls your existing src/main.py for RVC conversion.
    Returns: (audio_filepath_or_None, status_message, gr.update for download)
    """
    try:
        if not audio:
            return None, "❌ Upload audio file first", gr.update(visible=False, value=None)
        if not model or model == "No models":
            return None, "❌ No model selected", gr.update(visible=False, value=None)

        cmd = [
            "python", "src/main.py",
            "-i", str(audio), "-dir", str(model), "-p", str(pitch), "-k",
            "-ir", str(idx_rate), "-fr", str(flt_rad), "-rms", str(rms),
            "-palgo", str(algo), "-hop", "64", "-pro", str(prot),
            "-mv", str(mvol), "-bv", str(bvol), "-iv", str(ivol),
            "-pall", str(pitch_all), "-rsize", str(rev_sz),
            "-rwet", str(rev_wet), "-rdry", str(rev_dry),
            "-rdamp", str(rev_damp), "-oformat", str(fmt)
        ]

        work_dir = '/content/FIX'
        if not os.path.exists(work_dir):
            work_dir = os.getcwd()

        proc = subprocess.Popen(cmd, stdout=subprocess.PIPE, stderr=subprocess.STDOUT, universal_newlines=True, cwd=work_dir)
        # stream logs to notebook output for debugging
        for line in proc.stdout:
            print(line, end='')
        proc.wait()

        out_dir = Path(work_dir) / 'song_output'
        if out_dir.exists():
            files = list(out_dir.rglob(f'*.{fmt}'))
            if files:
                latest = max(files, key=os.path.getctime)
                return str(latest), "✅ Conversion complete!", gr.update(visible=True, value=str(latest))

        return None, "❌ Output not found. Check logs in the notebook.", gr.update(visible=False, value=None)

    except Exception as e:
        return None, f"❌ Error: {str(e)}", gr.update(visible=False, value=None)

# ------------------------- Text-to-Speech (Murf REST API) -------------------
MURF_API_URL = "https://api.murf.ai/v1/speech/generate"

def _download_audio_from_url(audio_url, out_path):
    resp = requests.get(audio_url, stream=True, timeout=60)
    if resp.status_code != 200:
        raise RuntimeError(f"Download failed: {resp.status_code} - {resp.text}")
    with open(out_path, 'wb') as f:
        for chunk in resp.iter_content(chunk_size=8192):
            if chunk:
                f.write(chunk)
    return out_path

def text_to_speech(text, voice, fmt, rate, channel):
    """
    Calls Murf API and saves output to /content/FIX/tts_output.<fmt>
    Returns: (audio_filepath_or_None, status_message, gr.update for download)
    """
    try:
        if not MURF_API_KEY or not MURF_API_KEY.strip():
            return None, "❌ Add Murf API key in Cell 7 and rerun", gr.update(visible=False, value=None)
        if not text or not text.strip():
            return None, "❌ Enter text first", gr.update(visible=False, value=None)

        headers = {"Content-Type": "application/json", "api-key": MURF_API_KEY.strip()}
        payload = {
            "text": text,
            "voiceId": voice,
            "format": fmt.lower(),
            "sampleRate": int(rate),
            "channelType": channel.lower()
        }

        resp = requests.post(MURF_API_URL, headers=headers, data=json.dumps(payload), timeout=60)
        if resp.status_code == 200:
            data = resp.json()
            audio_url = (data.get('audioFile') or data.get('audio_file') or data.get('audioUrl') or data.get('audio_url') or data.get('url'))
            if not audio_url:
                # If API returns base64 or a different key, surface a helpful message
                return None, f"❌ API returned 200 but no audio URL found: {json.dumps(data)[:300]}", gr.update(visible=False, value=None)
            out_path = f"/content/FIX/tts_output.{fmt.lower()}"
            _download_audio_from_url(audio_url, out_path)
            return out_path, "✅ Speech generated!", gr.update(visible=True, value=out_path)
        try:
            err = resp.json()
        except Exception:
            err = resp.text
        return None, f"❌ Murf API error ({resp.status_code}): {err}", gr.update(visible=False, value=None)

    except Exception as e:
        return None, f"❌ Error: {str(e)}", gr.update(visible=False, value=None)

# ------------------------- Gradio UI Setup ----------------------------------
models = get_available_models()
voices = [
    "en-US-terrell", "en-US-natalie", "en-US-julia", "en-US-wayne",
    "en-US-charles", "en-UK-hazel", "en-AU-kylie", "en-IN-aarav",
    "en-IN-priya", "es-ES-elvira", "fr-FR-adelie", "de-DE-matthias"
]

# Local uploaded images (developer-provided) — using exact local paths so Gradio displays them.
img1 = "/content/image.png"


# ------------------------- Futuristic CSS ----------------------------------
custom_css = r"""
/* Fonts */
@import url('https://fonts.googleapis.com/css2?family=Orbitron:wght@400;700&family=Space+Grotesk:wght@300;400;600;700&display=swap');

:root{
  --bg-1: #020205;
  --neon-green: #00ff9c;
  --neon-cyan: #00c2ff;
  --neon-magenta: #ff4dd2;
  --muted: rgba(255,255,255,0.78);
  --card: rgba(255,255,255,0.02);
  --glass: rgba(255,255,255,0.03);
  --radius: 14px;
}

/* Background: moving matrix-like gradient & faint grid */
body, .gradio-container {
  background: radial-gradient(circle at 10% 10%, rgba(0,194,255,0.03), transparent 8%),
              radial-gradient(circle at 90% 90%, rgba(255,77,210,0.02), transparent 10%),
              linear-gradient(180deg, #020205 0%, #071026 60%, #05040a 100%);
  color: white;
  font-family: 'Space Grotesk', 'Orbitron', system-ui, -apple-system, 'Segoe UI', Roboto, Arial;
  -webkit-font-smoothing:antialiased; -moz-osx-font-smoothing:grayscale;
  overflow-x: hidden;
}

/* subtle animated vertical lines to get matrix vibe */
@keyframes lineMove {
  0% { background-position: 0 0, 0 0; }
  50% { background-position: -40px 0, 40px 0; }
  100% { background-position: 0 0, 0 0; }
}
.gradio-container::after {
  content: "";
  position: absolute;
  inset: 0;
  background-image: linear-gradient(transparent, rgba(255,255,255,0.01) 50%, transparent),
                    linear-gradient(90deg, rgba(0,255,150,0.02) 0px, transparent 60px);
  pointer-events: none;
  mix-blend-mode:overlay;
  animation: lineMove 12s linear infinite;
  z-index:0;
}

/* NAV (tabs styled as horizontal navbar) */
.gradio-tabs .tabs {
  display:flex; justify-content:center; gap:12px; align-items:center;
  padding:12px 18px; border-radius:12px; margin-bottom:18px;
  background: linear-gradient(180deg, rgba(255,255,255,0.01), rgba(255,255,255,0.00));
  z-index:5;
}
.gradio-tabs .tab {
  padding:10px 14px; border-radius:10px; font-weight:700; color:var(--muted);
  letter-spacing:0.3px;
  transition: transform .16s ease, color .12s ease, box-shadow .12s;
  border:1px solid transparent;
  background: linear-gradient(180deg, rgba(255,255,255,0.01), rgba(255,255,255,0.00));
}
.gradio-tabs .tab:hover { transform: translateY(-4px); color: white; box-shadow: 0 10px 30px rgba(0,194,255,0.04); }
.gradio-tabs .tab--selected {
  background: linear-gradient(90deg, rgba(0,194,255,0.08), rgba(255,77,210,0.06));
  color: white !important;
  box-shadow: 0 18px 60px rgba(0,194,255,0.06);
}

/* Hero / title */
.header-row { display:flex; justify-content:space-between; align-items:center; gap:12px; margin-bottom:6px; z-index:4; }
.title-large { font-size:34px; font-weight:800; letter-spacing:-0.6px; text-shadow: 0 0 20px rgba(0,194,255,0.08), 0 0 42px rgba(255,77,210,0.04); }
.subtitle { color: rgba(255,255,255,0.85); font-size:15px; margin-top:6px; }

/* Neon tag */
.neon-tag { display:inline-block; padding:6px 10px; border-radius:999px; font-weight:700; color:#001219;
 background: linear-gradient(90deg, rgba(0,194,255,0.9), rgba(255,77,210,0.8)); box-shadow: 0 12px 40px rgba(0,194,255,0.08); }

/* 3D cards grid */
.grid-3d { display:grid; grid-template-columns: repeat(3, 1fr); gap:14px; margin-top:16px; }
.card-3d {
  background: linear-gradient(180deg, rgba(255,255,255,0.02), rgba(255,255,255,0.01));
  border-radius: 14px; padding:18px; border: 1px solid rgba(0,255,150,0.02);
  box-shadow: 0 30px 80px rgba(2,6,23,0.7);
  transform-style: preserve-3d;
  transition: transform 0.35s cubic-bezier(.2,.9,.2,1), box-shadow .25s;
  overflow:hidden;
}
.card-3d:hover { transform: translateY(-8px) rotateX(0.8deg) translateZ(6px); box-shadow: 0 40px 120px rgba(8,10,40,0.9); }
.card-title { font-weight:700; font-size:16px; color: #fff; }
.card-desc { color: rgba(255,255,255,0.78); margin-top:8px; font-size:13px; }

/* cyberpunk box style */
.cyber-box {
  border-radius:12px; padding:12px; background: linear-gradient(90deg, rgba(0,255,150,0.03), rgba(0,194,255,0.02));
  border: 1px solid rgba(0,255,150,0.06); box-shadow: 0 8px 30px rgba(0,255,150,0.03);
}

/* big glowing section headings */
.section-heading {
  font-weight:800; font-size:20px; color: #fff; padding-bottom:8px;
  text-shadow: 0 0 18px rgba(0,194,255,0.06);
  border-bottom: 1px solid rgba(255,255,255,0.03); margin-top:18px;
}

/* audio player and download file */
audio { width:100% !important; border-radius:8px; }
.gr-file { margin-top:8px; }

/* form inputs styling (no pointer blocking) */
input, textarea, select, .gradio-input, .gradio-dropdown, .gradio-radio {
  background: rgba(255,255,255,0.02) !important; border-radius:10px !important; color:#fff !important;
  border:1px solid rgba(255,255,255,0.03) !important;
}

/* radio fix visibility */
.gradio-radio input[type="radio"] {
  appearance: auto !important;
  accent-color: var(--neon-cyan) !important;
  width:18px !important; height:18px !important; cursor:pointer;
}
.gradio-radio label { cursor:pointer !important; display:flex; align-items:center; gap:8px; }

/* footer removal helper - hide default gradio footer items if present */
footer, .gradio-root .footer { display:none !important; }

/* responsive */
@media (max-width:1000px) { .grid-3d { grid-template-columns: 1fr 1fr; } .card-3d{ padding:12px; } }
@media (max-width:640px) { .grid-3d { grid-template-columns: 1fr; } .header-row{ flex-direction:column; align-items:flex-start; gap:10px; } }
"""

# ---------------------- Build Gradio App ------------------------------------
with gr.Blocks(title="Voice AI Studio — Futuristic", css=custom_css) as app:
    # Top header (title + small neon tag)
    with gr.Row(elem_classes="header-row"):
        with gr.Column(scale=1):
            gr.HTML(f"""
              <div style="display:flex; gap:12px; align-items:center;">
                <div class="neon-tag">Studio • Beta</div>
                <div style="display:flex; flex-direction:column;">
                  <div class="title-large">Voice AI Studio</div>
                  <div class="subtitle">Futuristic voice conversion & hyper-real TTS — cyberpunk UI, studio results.</div>
                </div>
              </div>
            """)
        with gr.Column(scale=0.3, min_width=160):
            # hide timestamp and other clutter; show small minimal links
            gr.HTML(f"<div style='text-align:right; color:rgba(255,255,255,0.6); font-size:13px;'>Ultra-Premium Dashboard</div>")

    # Tabs used as navbar
    with gr.Tabs():
        # HOME - long landing with many sections
        with gr.TabItem("Home"):
            # Hero: left text, right image collage
            with gr.Row():
                with gr.Column(scale=1):
                    gr.HTML("""
                      <div style="margin-bottom:8px;"><span class="section-heading">Welcome</span></div>
                      <h1 style="font-size:28px; margin:6px 0; font-weight:800; letter-spacing:-0.6px; color:#E6FFFF;">Create. Convert. Captivate.</h1>
                      <p style="color:rgba(255,255,255,0.85); font-size:15px; max-width:780px;">This futuristic dashboard brings studio-grade voice conversion and high-fidelity TTS into a single, beautiful interface — optimized for creators and developers. Explore features, tweak audio, and download high-quality outputs with a single click.</p>
                      <div style="margin-top:12px; display:flex; gap:10px;">
                        <button class="gr-button" onclick="document.querySelectorAll('[role=tablist] button')[1].click();">Open Voice Changer</button>
                        <button class="gr-button" style="background:transparent; border:1px solid rgba(0,255,150,0.06);">Explore Features</button>
                      </div>
                    """)
                with gr.Column(scale=0.8):
                    # show the three uploaded images as a small collage (paths provided)
                    try:
                        # show using HTML img tags (Gradio will display local images)
                        gr.Image(value="/content/image.png", label="", show_download_button=False, show_label=False)

                    except Exception:
                        pass

            # 3D Features grid
            gr.HTML('<div class="section-heading">Core Features</div>')
            with gr.Row():
                gr.HTML('<div class="grid-3d" style="width:100%;">'
                        f'<div class="card-3d"><div class="card-title">Studio-grade Conversion</div><div class="card-desc">High-quality RVC models with granular pitch and reverb controls.</div></div>'
                        f'<div class="card-3d"><div class="card-title">Hyper-real TTS</div><div class="card-desc">Murf-backed voices, multi-sample rate support and stereo output.</div></div>'
                        f'<div class="card-3d"><div class="card-title">Fast Workflow</div><div class="card-desc">Upload → tweak settings → convert → download — optimized for creators.</div></div>'
                        '</div>')

            # Technology & Info (cyberpunk boxes)
            gr.HTML('<div class="section-heading">Technology</div>')
            with gr.Row():
                gr.HTML('<div style="display:flex; gap:12px; flex-wrap:wrap;">'
                        '<div class="cyber-box" style="min-width:280px;"><strong>RVC Models</strong><div style="margin-top:6px; color:rgba(255,255,255,0.85);">State-of-the-art voice conversion using pre-trained RVC models.</div></div>'
                        '<div class="cyber-box" style="min-width:280px;"><strong>Murf TTS</strong><div style="margin-top:6px; color:rgba(255,255,255,0.85);">High-fidelity REST TTS integration with multiple sample rates.</div></div>'
                        '<div class="cyber-box" style="min-width:280px;"><strong>Export & Download</strong><div style="margin-top:6px; color:rgba(255,255,255,0.85);">Automatic download widget appears when audio is ready.</div></div>'
                        '</div>')

            # Quote block (stylish)
            gr.HTML('<div style="margin-top:18px; padding:18px; border-radius:12px; background: linear-gradient(90deg, rgba(0,194,255,0.04), rgba(255,77,210,0.03)); border:1px solid rgba(0,194,255,0.06);">'
                    '<div style="font-weight:800; font-size:18px; color:#E8FFFF;">“Voice is the next frontier — craft it with precision.”</div>'
                    '<div style="margin-top:8px; color:rgba(255,255,255,0.8);">— Voice AI Studio</div>'
                    '</div>')

            # Long info boxes area (replace default help text — stylish boxes)
            gr.HTML('<div class="section-heading">How to use</div>')
            gr.HTML('''
              <div style="display:grid; grid-template-columns: repeat(2,1fr); gap:12px; margin-top:8px;">
                <div class="card-3d">
                  <div style="font-weight:700;">Voice Changer — Quick Steps</div>
                  <ol style="margin-top:8px; color:rgba(255,255,255,0.85);">
                    <li>Upload audio (WAV recommended).</li>
                    <li>Select a model and format.</li>
                    <li>Adjust pitch, index and advanced effects.</li>
                    <li>Click Convert. Output saved to <code style="color:#00ff9c;">/content/FIX/song_output</code>.</li>
                    <li>Use the player to preview and the Download widget to save.</li>
                  </ol>
                </div>
                <div class="card-3d">
                  <div style="font-weight:700;">Text to Speech — Quick Steps</div>
                  <ol style="margin-top:8px; color:rgba(255,255,255,0.85);">
                    <li>Enter text and choose voice and format.</li>
                    <li>Select sample rate and channel.</li>
                    <li>Click Generate. Output saved to <code style="color:#00ff9c;">/content/FIX/tts_output.*</code>.</li>
                    <li>Preview and download when ready.</li>
                  </ol>
                </div>
              </div>
            ''')

            # Additional sections to make page long and rich
            gr.HTML('<div class="section-heading">Roadmap</div>')
            gr.HTML('''
              <div style="display:flex; gap:12px; margin-top:8px; flex-wrap:wrap;">
                <div class="card-3d" style="min-width:260px;"><strong>Realtime preview</strong><div style="color:rgba(255,255,255,0.85); margin-top:6px;">In-browser real-time playback</div></div>
                <div class="card-3d" style="min-width:260px;"><strong>Batch processing</strong><div style="color:rgba(255,255,255,0.85); margin-top:6px;">Queue multiple files for conversion</div></div>
                <div class="card-3d" style="min-width:260px;"><strong>Presets</strong><div style="color:rgba(255,255,255,0.85); margin-top:6px;">One-click style presets</div></div>
              </div>
            ''')

        # VOICE CHANGER - functionality preserved
        with gr.TabItem("Voice Changer"):
            with gr.Row():
                with gr.Column(scale=1):
                    vc_audio = gr.Audio(label="Upload Audio", type="filepath")
                    vc_model = gr.Dropdown(choices=models, label="Model", value=(models[0] if models else "No models"))
                    vc_pitch = gr.Slider(-12, 12, 0, step=1, label="Pitch")
                    vc_idx = gr.Slider(0, 1, 0.75, step=0.01, label="Index Rate")
                    vc_flt = gr.Slider(0, 7, 3, step=1, label="Filter Radius")
                    with gr.Accordion("Advanced", open=False):
                        vc_rms = gr.Slider(0, 1, 0.25, step=0.01, label="RMS Mix")
                        vc_prot = gr.Slider(0, 0.5, 0.33, step=0.01, label="Protect")
                        vc_mvol = gr.Slider(-20, 20, 0, step=1, label="Main Vol")
                        vc_bvol = gr.Slider(-20, 20, 0, step=1, label="Backup Vol")
                        vc_ivol = gr.Slider(-20, 20, 0, step=1, label="Inst Vol")
                        vc_pall = gr.Slider(-12, 12, 0, step=1, label="Pitch All")
                        vc_rsize = gr.Slider(0, 1, 0.15, step=0.01, label="Reverb Size")
                        vc_rwet = gr.Slider(0, 1, 0.2, step=0.01, label="Reverb Wet")
                        vc_rdry = gr.Slider(0, 1, 0.8, step=0.01, label="Reverb Dry")
                        vc_rdamp = gr.Slider(0, 1, 0.7, step=0.01, label="Reverb Damp")
                        vc_algo = gr.Radio(["rmvpe", "mangio-crepe"], value="rmvpe", label="Algorithm")
                        vc_fmt = gr.Radio(["wav", "mp3"], value="wav", label="Format")
                    vc_btn = gr.Button("Convert", variant="primary")

                with gr.Column(scale=1):
                    vc_out = gr.Audio(label="Converted Audio", type="filepath")
                    vc_msg = gr.Textbox(label="Status", lines=2)
                    vc_download = gr.File(label="Download Audio", visible=False)

            vc_btn.click(voice_conversion,
                         inputs=[vc_audio, vc_model, vc_pitch, vc_idx, vc_flt, vc_rms,
                                 vc_prot, vc_mvol, vc_bvol, vc_ivol, vc_pall, vc_rsize,
                                 vc_rwet, vc_rdry, vc_rdamp, vc_fmt, vc_algo],
                         outputs=[vc_out, vc_msg, vc_download])

        # TEXT TO SPEECH - functionality preserved
        with gr.TabItem("Text to Speech"):
            with gr.Row():
                with gr.Column(scale=1):
                    tts_text = gr.Textbox(label="Text", lines=10, placeholder="Enter text here...")
                    tts_voice = gr.Dropdown(choices=voices, label="Voice", value="en-US-terrell")
                    tts_fmt = gr.Radio(["WAV", "MP3", "FLAC"], value="WAV", label="Format")
                    tts_rate = gr.Radio(["8000", "24000", "44100", "48000"], value="24000", label="Sample Rate")
                    tts_ch = gr.Radio(["MONO", "STEREO"], value="STEREO", label="Channel")
                    tts_btn = gr.Button("Generate", variant="primary")

                with gr.Column(scale=1):
                    tts_out = gr.Audio(label="Generated Audio", type="filepath")
                    tts_msg = gr.Textbox(label="Status", lines=2)
                    tts_download = gr.File(label="Download Audio", visible=False)
                    gr.Markdown("""
                    <div style="color:rgba(255,255,255,0.78); margin-top:6px;">Voice Guide — US: Terrell, Natalie, Julia, Wayne • UK: Charles, Hazel • India: Aarav, Priya</div>
                    """)

            tts_btn.click(text_to_speech,
                          inputs=[tts_text, tts_voice, tts_fmt, tts_rate, tts_ch],
                          outputs=[tts_out, tts_msg, tts_download])

        # ABOUT - premium styled about section (replace previous text)
        with gr.TabItem("About"):
            gr.HTML('<div class="section-heading">About Voice AI Studio (Futuristic)</div>')
            gr.HTML('''
              <div style="display:grid; grid-template-columns: 1fr 380px; gap:12px; align-items:start;">
                <div class="card-3d">
                  <h3 style="margin:0 0 8px 0; font-weight:800;">Vision</h3>
                  <p style="color:rgba(255,255,255,0.85); margin:0 0 10px 0;">This workspace brings studio-grade voice conversion and high-fidelity TTS into a single, beautiful interface — running inside Colab. Our focus: quality-first audio, rapid iteration, and a modern UI that inspires creativity.</p>
                  <div style="margin-top:10px;">
                    <div style="font-weight:700;">Design</div><div style="color:rgba(255,255,255,0.85); margin-top:6px;">Multi-gradient, animated, futuristic UI with 3D cards and neon accents.</div>
                    <div style="font-weight:700; margin-top:8px;">Functionality</div><div style="color:rgba(255,255,255,0.85); margin-top:6px;">RVC-based conversion + Murf-powered TTS; simple steps to generate high-quality audio.</div>
                    <div style="font-weight:700; margin-top:8px;">Privacy</div><div style="color:rgba(255,255,255,0.85); margin-top:6px;">Keep API keys private when sharing notebooks. Remove keys before public sharing.</div>
                  </div>
                </div>
                <div class="card-3d">
                  <h4 style="margin:0 0 8px 0; font-weight:800;">Inspiration Quote</h4>
                  <div style="font-size:16px; font-weight:700; color:#E8FFFF;">“Sound is sculpture — shape it.”</div>
                  <div style="margin-top:12px; color:rgba(255,255,255,0.85);">Built for creators who demand premium audio quality and modern workflows.</div>
                </div>
              </div>
            ''')

        # HELP - replace with long stylish help/info page (user requested removal of raw available models text)
        with gr.TabItem("Help"):
            gr.HTML('<div class="section-heading">Quick Help & Guidance</div>')
            gr.HTML('''
              <div style="display:grid; grid-template-columns: 1fr; gap:12px;">
                <div class="card-3d">
                  <h4 style="margin:0 0 8px 0; font-weight:800;">Voice Changer</h4>
                  <ol style="color:rgba(255,255,255,0.85); margin:0; padding-left:16px;">
                    <li>Upload your audio (WAV recommended for best fidelity).</li>
                    <li>Select the voice model and desired format (WAV/MP3).</li>
                    <li>Tweak pitch, index rate, and advanced options as needed.</li>
                    <li>Click <strong>Convert</strong>. The converted file will be saved to <code style="color:#00ff9c;">/content/FIX/song_output</code>.</li>
                    <li>Preview using the player and click <strong>Download Audio</strong> (appears when conversion completes).</li>
                  </ol>
                </div>
                <div class="card-3d">
                  <h4 style="margin:0 0 8px 0; font-weight:800;">Text to Speech (TTS)</h4>
                  <ol style="color:rgba(255,255,255,0.85); margin:0; padding-left:16px;">
                    <li>Enter the text you want to convert to speech.</li>
                    <li>Choose a voice, format, sample rate and channel.</li>
                    <li>Click <strong>Generate</strong>. The TTS will be saved to <code style="color:#00ff9c;">/content/FIX/tts_output.*</code>.</li>
                    <li>Use the player to preview and the download widget to get the file locally.</li>
                  </ol>
                </div>
                <div class="card-3d">
                  <h4 style="margin:0 0 8px 0; font-weight:800;">Troubleshooting</h4>
                  <ul style="color:rgba(255,255,255,0.85); margin:0; padding-left:16px;">
                    <li>If audio does not play, run <code>!ls -lh /content/FIX/song_output</code> to verify files exist and are non-empty.</li>
                    <li>If Murf returns no URL, paste the raw API response and I will add base64 handling.</li>
                    <li>Ensure your models are in the folder <code>/content/FIX/rvc_models</code>.</li>
                  </ul>
                </div>
              </div>
            ''')

    # Footer area — minimal (removed default Gradio footer and timestamp)
    gr.HTML(f"<div style='margin-top:22px; text-align:center; color:rgba(255,255,255,0.6); font-size:13px;'>Voice AI Studio — Ultra-Futuristic • Powered in Colab</div>")

# Launch app
print("Launching Futuristic Dashboard UI...")
app.launch(share=True, debug=True)
print("UI launched.")

```
